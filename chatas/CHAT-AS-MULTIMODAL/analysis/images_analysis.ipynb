```python
from typing import Dict, List, Tuple
import pandas as pd
def create_image_path_by_url(
    image_names_dir: str, images_dir: str
) -> Tuple[Dict[str, str], Dict[str, str]]:
    image_path_by_url = {}
    url_by_image_path = {}
    for file in os.listdir(image_names_dir):
        with open(os.path.join(image_names_dir, file)) as f:
            image_names = pd.read_csv(f, sep="\t", header=None)
            for i in range(len(image_names)):
                if '106868' in str(image_names.iloc[i, 1]):
                    print(file)
                image_path_by_url[image_names.iloc[i, 0]] = os.path.join(
                    images_dir, str(image_names.iloc[i, 1])
                )
                url_by_image_path[os.path.join(images_dir, str(image_names.iloc[i, 1]))] = image_names.iloc[i, 0]
    return image_path_by_url, url_by_image_path
```

```python
import os
import requests
import shutil
from requests.adapters import HTTPAdapter
from requests.packages.urllib3.util.retry import Retry

# Ensure the directory exists
os.makedirs('./defective_images', exist_ok=True)

# Set up retries for requests in case of connection issues
retry_strategy = Retry(
    total=5,  # Retry up to 5 times
    backoff_factor=1,  # Wait progressively longer after each retry (1s, 2s, 4s, etc.)
    status_forcelist=[429, 500, 502, 503, 504]  # Retry on these HTTP status codes
)
adapter = HTTPAdapter(max_retries=retry_strategy)
http = requests.Session()
http.mount("http://", adapter)
http.mount("https://", adapter)

# Custom headers to bypass potential 403 errors
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3"
}

# Download images
for ind, i in enumerate(defective_links):
    url = i
    name = defective_images[ind]
    
    try:
        # Make the request with headers
        response = http.get(url, headers=headers, stream=True, timeout=10)
        response.raise_for_status()  # Raise an error for bad status codes

        # Write the image to file
        with open(f'./defective_images/{name}', 'wb') as out_file:
            shutil.copyfileobj(response.raw, out_file)
        print(f'Downloaded: ./defective_images/{name}')
        
    except requests.exceptions.HTTPError as http_err:
        if response.status_code == 403:
            print(f"Failed to download {url}: 403 Forbidden. Possibly restricted access.")
        elif response.status_code == 404:
            print(f"Failed to download {url}: 404 Not Found. The resource might not exist.")
        elif response.status_code == 400:
            print(f"Failed to download {url}: 400 Bad Request. There may be an issue with the URL.")
        else:
            print(f"Failed to download {url}: {http_err}")
    
    except requests.exceptions.RequestException as e:
        print(f"Failed to download {url}: {e}")
    
    finally:
        del response
```
